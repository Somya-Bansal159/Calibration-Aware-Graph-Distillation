```python
def combined(acc, ece):
    return (2 * acc * (1-ece)) / (acc + (1-ece))
```

```python
def make_split_cmap(base_hex):
    """Create a colormap:  light->base for values ≤1, base->dark for values >1"""
    rgb = np.array(to_rgb(base_hex))
    light = to_hex(rgb * 0.3 + np.array([1,1,1]) * 0.7)
    dark  = to_hex(np.clip(rgb * 0.5, 0, 1))
    cmap = LinearSegmentedColormap.from_list('split', [light, base_hex, dark])
    return cmap
```

```python
def create_tick_mesh1(xc, yc, zc, size):

    verts = np.array([
        [xc, yc + size, zc],
        [xc, yc + size/2, zc - size],
        [xc, yc - size, zc + size]
    ])

    return go.Mesh3d(
        x=verts[:,0],
        y=verts[:,1],
        z=verts[:,2],
        i=[0],
        j=[1],
        k=[2],
        color='black',
        showscale=False
    )

def create_tick_mesh2(xc, yc, zc, size):

    # Simple 3D "V" shaped tick in the XY plane
    verts = np.array([
        [xc, yc - size, zc],
        [xc, yc - size/2, zc - size],
        [xc, yc + size, zc + size]
    ])

    return go.Mesh3d(
        x=verts[:,0],
        y=verts[:,1],
        z=verts[:,2],
        i=[0],
        j=[1],
        k=[2],
        color='black',
        showscale=False
    )

def create_tick_mesh3(xc, yc, zc, size):

    # Simple 3D "V" shaped tick in the XY plane
    verts = np.array([
        [xc, yc + size, zc],
        [xc, yc + size/2, zc - size],
        [xc, yc + 5*size, zc + 2*size]
    ])

    return go.Mesh3d(
        x=verts[:,0],
        y=verts[:,1],
        z=verts[:,2],
        i=[0],
        j=[1],
        k=[2],
        color='black',
        showscale=False
    )

def create_tick_mesh4(xc, yc, zc, size):

    # Simple 3D "V" shaped tick in the XY plane
    verts = np.array([
        [xc, yc - size, zc],
        [xc, yc - size/2, zc - size],
        [xc, yc - 5*size, zc + 2*size]
    ])

    return go.Mesh3d(
        x=verts[:,0],
        y=verts[:,1],
        z=verts[:,2],
        i=[0],
        j=[1],
        k=[2],
        color='black',
        showscale=False
    )

def plot_vertical_plane_heatmap_with_ticks(R, C, fig, data, X, base_hex, norm, size):
    """
    Plot a VERTICAL plane at X with color mapping.  
    Cells with values > 1 get tick (checkmark) markers.
    """
    cmap = make_split_cmap(base_hex)
    x = np.linspace(0, C, C+1)
    y = np.linspace(0, R, R+1)

    vertices = []
    for yi in y:
        for xi in x:
            vertices.append([xi, yi, X])

    vertices = np.array(vertices)

    I = []
    J = []
    K = []
    face_colors = []

    def vert_index(i, j):
        return i * (C+1) + j

    for i in range(R):
        for j in range(C):
            v0 = vert_index(i, j)
            v1 = vert_index(i, j+1)
            v2 = vert_index(i+1, j)
            v3 = vert_index(i+1, j+1)

            # first triangle
            I.append(v0)
            J.append(v1)
            K.append(v2)

            # second triangle
            I.append(v2)
            J.append(v1)
            K.append(v3)

            # same color for both triangles
            color = to_hex(cmap(norm(data[i, j])))
            # print(cmap(norm(data[i, j])), cmap(norm(data[i, j]))*2)
            face_colors.extend([color, color])


    fig.add_trace(go.Mesh3d(
        x=vertices[:,2] ,
        y=vertices[:,0],
        z=vertices[:,1],
        i=K,
        j=I,
        k=J,
        facecolor=face_colors,
        showscale=False,
        lighting=dict(
            ambient=0.8,
            diffuse=0.9,
            roughness=0.5,
            specular=1.0,
            fresnel=0.2
        ),
        lightposition=dict(
            x=100,
            y=100,
            z=100
        )
    ))

    rows, cols = data.shape

    for i in range(rows):
        for j in range(cols):
            if data[i, j] > 1:

                # center of the cell
                xc = j + 0.5
                yc = i + 0.5

                fig.add_trace(create_tick_mesh1(X+0.01, xc+0.1, yc, size))
                fig.add_trace(create_tick_mesh2(X-0.01, xc+0.1, yc, size))

                fig.add_trace(create_tick_mesh3(X+0.01, xc+0.1, yc, size))
                fig.add_trace(create_tick_mesh4(X-0.01, xc+0.1, yc, size))

def camera_from_angles(elev_deg, azim_deg, zoom):

    elev = np.radians(elev_deg)
    azim = np.radians(azim_deg)

    x = zoom * np.cos(elev) * np.sin(azim)
    y = zoom * np.cos(elev) * np.cos(azim)
    z = zoom * np.sin(elev)

    return dict(eye=dict(x=x, y=y, z=z), projection=dict(type="orthographic"))
```
