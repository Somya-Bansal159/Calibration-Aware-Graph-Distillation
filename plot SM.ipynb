```python
def combined(acc, ece):
    return (2 * acc * (1-ece)) / (acc + (1-ece))
```

```python
def make_split_cmap(base_hex):
    """Create a colormap:  light->base for values ≤1, base->dark for values >1"""
    rgb = np.array(to_rgb(base_hex))
    light = to_hex(rgb * 0.3 + np.array([1,1,1]) * 0.7)
    dark  = to_hex(np.clip(rgb * 0.5, 0, 1))
    cmap = LinearSegmentedColormap.from_list('split', [light, base_hex, dark])
    return cmap
```

```python
def plot_vertical_plane_heatmap_with_ticks(ax, Y, Z, data, X, base_hex, norm, alpha=0.9):
    """
    Plot a VERTICAL plane at X with color mapping.  
    Cells with values > 1 get tick (checkmark) markers.
    """
    cmap = make_split_cmap(base_hex)
    X_plane = np.full_like(Y, X, dtype=float)
    colors_grid = cmap(norm(data))
    
    ax.plot_surface(X_plane, Y, Z,
                   facecolors=colors_grid,
                   rstride=1, cstride=1,
                   linewidth=0, antialiased=False,
                   shade=False, alpha=alpha)
    
    # Mark cells where value > 1 with TICK (checkmark) markers and borders
    rows, cols = data.shape
    for i in range(rows):
        for j in range(cols):
            if data[i, j] > 1:
                y_center = Y[i, j] + 0.5
                z_center = Z[i, j] + 0.5
                
                # Add TICK (checkmark) marker at cell center
                ax.text(X, y_center, z_center, '✓',
                       fontsize=12, fontweight='bold', color='black',
                       ha='center', va='center', zorder=10)
                
                # Add cell border (rectangle outline)
                y_corners = [Y[i, j], Y[i, j] + 1, Y[i, j] + 1, Y[i, j], Y[i, j]]
                z_corners = [Z[i, j], Z[i, j], Z[i, j] + 1, Z[i, j] + 1, Z[i, j]]
                x_corners = [X] * 5
                ax.plot(x_corners, y_corners, z_corners, 
                       color='black', linewidth=1.5, zorder=10)
```
